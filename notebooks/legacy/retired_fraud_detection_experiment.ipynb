```python
# ============================================================
# COMPLETE FRAUD FEATURE ENGINEERING + VALIDATION PIPELINE
# ============================================================
#
# SINGLE SOURCE OF TRUTH:
#       clean_df
#
# OUTPUT:
#       engineered_df
#
# IMPORTANT LEAKAGE RULE:
#
# For a transaction occurring at step t,
# historical features may only use:
#
#       step < t
#
# Transactions at the SAME step are never included in history.
#
# ============================================================


import pandas as pd
import numpy as np


# ============================================================
# 0. SOURCE DATA
# ============================================================

# NEVER modify clean_df directly.
# Everything starts from this copy.

source_df = (
    clean_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Required columns
# ------------------------------------------------------------

required_columns = [
    "step",
    "customer",
    "age",
    "gender",
    "merchant",
    "category",
    "amount",
    "fraud"
]


missing_required = [
    col
    for col in required_columns
    if col not in source_df.columns
]


if missing_required:
    raise ValueError(
        f"Missing required columns: {missing_required}"
    )


# ------------------------------------------------------------
# Normalize string/categorical columns ONCE
# ------------------------------------------------------------
#
# This prevents the earlier problem where:
#
# clean_df:
#     C123
#
# engineered_df:
#     'C123'
#
# ------------------------------------------------------------

string_columns = [
    "customer",
    "merchant",
    "category",
    "age",
    "gender"
]


for col in string_columns:

    source_df[col] = (
        source_df[col]
        .astype(str)
        .str.strip()
        .str.strip("'")
        .str.strip('"')
    )


# ------------------------------------------------------------
# Make sure numerical columns are numeric
# ------------------------------------------------------------

source_df["step"] = pd.to_numeric(
    source_df["step"],
    errors="raise"
)

source_df["amount"] = pd.to_numeric(
    source_df["amount"],
    errors="raise"
)

source_df["fraud"] = pd.to_numeric(
    source_df["fraud"],
    errors="raise"
)


# ------------------------------------------------------------
# Preserve original row order
# ------------------------------------------------------------

source_df["_row_id"] = np.arange(
    len(source_df)
)


print("=" * 70)
print("SOURCE DATA")
print("=" * 70)

print(f"Rows:    {source_df.shape[0]:,}")
print(f"Columns: {source_df.shape[1] - 1:,}")


# ============================================================
# 1. BASIC TIME / TRANSACTION FEATURES
# ============================================================

# BankSim definition: step is the zero-based simulated day index.
# The following two legacy fields reproduce the frozen baseline only;
# they are not valid calendar-day or hour-of-day features.
# See experiments/temporal_model_comparison.py before replacing artifacts.

source_df["day"] = (
    source_df["step"] // 24
)

source_df["hour_of_day"] = (
    source_df["step"] % 24
)

# Right-skewed amount transformation
source_df["log_amount"] = np.log1p(
    source_df["amount"]
)


# ============================================================
# 2. ENTITY HISTORY FUNCTION
# ============================================================
#
# Builds history for:
#
#   customer
#   merchant
#   category
#
# Includes:
#
#   previous transaction count
#   previous total amount
#   previous average
#   previous std
#   previous min/max
#   transaction rate per simulated day
#   amount rate per simulated day
#   time since previous transaction
#   1 / 3 / 6 / 24 simulated-day activity
#   velocity ratio
#   recent amount ratio
#
# ============================================================


def build_entity_history(
    source,
    entity_col,
    prefix,
    recent_windows=(1, 3, 6, 24)
):

    # --------------------------------------------------------
    # Entity-step transaction statistics
    # --------------------------------------------------------

    temp = source[
        [
            entity_col,
            "step",
            "amount"
        ]
    ].copy()


    temp["_amount_squared"] = (
        temp["amount"] ** 2
    )


    step_stats = (
        temp
        .groupby(
            [
                entity_col,
                "step"
            ],
            observed=True,
            sort=True
        )
        .agg(

            step_transaction_count=(
                "amount",
                "size"
            ),

            step_total_amount=(
                "amount",
                "sum"
            ),

            step_amount_squared_sum=(
                "_amount_squared",
                "sum"
            ),

            step_min_amount=(
                "amount",
                "min"
            ),

            step_max_amount=(
                "amount",
                "max"
            )
        )
        .reset_index()
        .sort_values(
            [
                entity_col,
                "step"
            ]
        )
        .reset_index(drop=True)
    )


    grouped = step_stats.groupby(
        entity_col,
        sort=False
    )


    # ========================================================
    # PREVIOUS TRANSACTION COUNT
    # ========================================================

    previous_count = (
        grouped[
            "step_transaction_count"
        ].cumsum()
        -
        step_stats[
            "step_transaction_count"
        ]
    )


    step_stats[
        f"{prefix}_previous_transaction_count"
    ] = previous_count


    # ========================================================
    # PREVIOUS TOTAL AMOUNT
    # ========================================================

    previous_total = (
        grouped[
            "step_total_amount"
        ].cumsum()
        -
        step_stats[
            "step_total_amount"
        ]
    )


    step_stats[
        f"{prefix}_previous_total_amount"
    ] = previous_total


    # ========================================================
    # PREVIOUS AVERAGE AMOUNT
    # ========================================================

    step_stats[
        f"{prefix}_previous_avg_amount"
    ] = (
        previous_total
        /
        previous_count.replace(
            0,
            np.nan
        )
    )


    # ========================================================
    # PREVIOUS STANDARD DEVIATION
    # ========================================================

    previous_squared_sum = (
        grouped[
            "step_amount_squared_sum"
        ].cumsum()
        -
        step_stats[
            "step_amount_squared_sum"
        ]
    )


    safe_count = previous_count.replace(
        0,
        np.nan
    )


    variance_numerator = (
        previous_squared_sum
        -
        (
            previous_total ** 2
            /
            safe_count
        )
    )


    # Protect against tiny negative floating-point values
    variance_numerator = (
        variance_numerator
        .clip(lower=0)
    )


    step_stats[
        f"{prefix}_previous_std_amount"
    ] = np.sqrt(
        variance_numerator
        /
        (
            previous_count - 1
        ).where(
            previous_count > 1
        )
    )


    # ========================================================
    # PREVIOUS MINIMUM
    # ========================================================

    step_stats["_shifted_min"] = (
        grouped[
            "step_min_amount"
        ].shift(1)
    )


    step_stats[
        f"{prefix}_previous_min_amount"
    ] = (
        step_stats
        .groupby(
            entity_col,
            sort=False
        )["_shifted_min"]
        .cummin()
    )


    # ========================================================
    # PREVIOUS MAXIMUM
    # ========================================================

    step_stats["_shifted_max"] = (
        grouped[
            "step_max_amount"
        ].shift(1)
    )


    step_stats[
        f"{prefix}_previous_max_amount"
    ] = (
        step_stats
        .groupby(
            entity_col,
            sort=False
        )["_shifted_max"]
        .cummax()
    )


    # ========================================================
    # TIME SINCE LAST TRANSACTION
    # ========================================================

    previous_step = grouped[
        "step"
    ].shift(1)


    step_stats[
        f"{prefix}_steps_since_last_transaction"
    ] = (
        step_stats["step"]
        -
        previous_step
    )


    # ========================================================
    # HISTORICAL ACTIVITY RATE
    # ========================================================

    first_seen_step = grouped[
        "step"
    ].transform("min")


    elapsed_steps = (
        step_stats["step"]
        -
        first_seen_step
    ).astype(float)


    # Historical number of transactions per simulated day
    step_stats[
        f"{prefix}_transactions_per_step"
    ] = (
        previous_count
        /
        elapsed_steps.where(
            elapsed_steps > 0
        )
    )


    # Historical money volume per simulated day
    step_stats[
        f"{prefix}_amount_per_step"
    ] = (
        previous_total
        /
        elapsed_steps.where(
            elapsed_steps > 0
        )
    )


    # ========================================================
    # RECENT WINDOWS
    # ========================================================

    min_step = int(
        source["step"].min()
    )

    max_step = int(
        source["step"].max()
    )


    all_steps = np.arange(
        min_step,
        max_step + 1
    )


    # --------------------------------------------------------
    # Count matrix
    #
    # rows = entity
    # columns = simulated day
    # --------------------------------------------------------

    count_matrix = (
        step_stats
        .pivot(
            index=entity_col,
            columns="step",
            values="step_transaction_count"
        )
        .reindex(
            columns=all_steps,
            fill_value=0
        )
        .fillna(0)
    )


    # --------------------------------------------------------
    # Amount matrix
    # --------------------------------------------------------

    amount_matrix = (
        step_stats
        .pivot(
            index=entity_col,
            columns="step",
            values="step_total_amount"
        )
        .reindex(
            columns=all_steps,
            fill_value=0
        )
        .fillna(0)
    )


    # --------------------------------------------------------
    # CRITICAL LEAKAGE CONTROL
    #
    # Shift one simulated day:
    #
    # Current step t only sees:
    #
    #       t-1
    #       t-2
    #       ...
    #
    # Never t itself.
    # --------------------------------------------------------

    shifted_counts = (
        count_matrix
        .shift(
            1,
            axis=1
        )
        .fillna(0)
    )


    shifted_amounts = (
        amount_matrix
        .shift(
            1,
            axis=1
        )
        .fillna(0)
    )


    recent_feature_series = []


    for window in recent_windows:

        # ----------------------------------------------------
        # Recent transaction count
        # ----------------------------------------------------

        recent_count = (
            shifted_counts
            .T
            .rolling(
                window=window,
                min_periods=1
            )
            .sum()
            .T
            .stack()
            .rename(
                f"{prefix}_transactions_last_{window}_steps"
            )
        )


        # ----------------------------------------------------
        # Recent amount
        # ----------------------------------------------------

        recent_amount = (
            shifted_amounts
            .T
            .rolling(
                window=window,
                min_periods=1
            )
            .sum()
            .T
            .stack()
            .rename(
                f"{prefix}_amount_last_{window}_steps"
            )
        )


        recent_feature_series.extend(
            [
                recent_count,
                recent_amount
            ]
        )


    recent_features = (
        pd.concat(
            recent_feature_series,
            axis=1
        )
        .reset_index()
    )


    # ========================================================
    # MERGE HISTORICAL + RECENT FEATURES
    # ========================================================

    generated_columns = [
        col
        for col in step_stats.columns
        if col.startswith(
            prefix + "_"
        )
    ]


    feature_table = (
        step_stats[
            [
                entity_col,
                "step"
            ]
            +
            generated_columns
        ]
        .merge(
            recent_features,
            on=[
                entity_col,
                "step"
            ],
            how="left",
            validate="one_to_one"
        )
    )


    # ========================================================
    # VELOCITY RATIO
    # ========================================================
    #
    # previous-day transactions
    # --------------------------
    # historical transactions/simulated day
    #
    # ========================================================

    feature_table[
        f"{prefix}_velocity_ratio"
    ] = (
        feature_table[
            f"{prefix}_transactions_last_1_steps"
        ]
        /
        feature_table[
            f"{prefix}_transactions_per_step"
        ].replace(
            0,
            np.nan
        )
    )


    # ========================================================
    # RECENT MONEY RATIO
    # ========================================================
    #
    # previous-day money
    # -------------------
    # historical money/simulated day
    #
    # ========================================================

    feature_table[
        f"{prefix}_recent_amount_ratio"
    ] = (
        feature_table[
            f"{prefix}_amount_last_1_steps"
        ]
        /
        feature_table[
            f"{prefix}_amount_per_step"
        ].replace(
            0,
            np.nan
        )
    )


    # ========================================================
    # NEW ENTITY
    # ========================================================

    feature_table[
        f"is_new_{prefix}"
    ] = (
        feature_table[
            f"{prefix}_previous_transaction_count"
        ] == 0
    ).astype(int)


    return feature_table


# ============================================================
# 3. BUILD CUSTOMER HISTORY
# ============================================================

customer_features = build_entity_history(
    source=source_df,
    entity_col="customer",
    prefix="customer"
)


# ============================================================
# 4. BUILD MERCHANT HISTORY
# ============================================================

merchant_features = build_entity_history(
    source=source_df,
    entity_col="merchant",
    prefix="merchant"
)


# ============================================================
# 5. BUILD CATEGORY HISTORY
# ============================================================

category_features = build_entity_history(
    source=source_df,
    entity_col="category",
    prefix="category"
)


# ============================================================
# 6. MERGE INDIVIDUAL HISTORIES
# ============================================================

engineered_df = (
    source_df
    .merge(
        customer_features,
        on=[
            "customer",
            "step"
        ],
        how="left",
        validate="many_to_one"
    )
    .merge(
        merchant_features,
        on=[
            "merchant",
            "step"
        ],
        how="left",
        validate="many_to_one"
    )
    .merge(
        category_features,
        on=[
            "category",
            "step"
        ],
        how="left",
        validate="many_to_one"
    )
)


# ============================================================
# 7. CURRENT AMOUNT VS HISTORICAL ENTITY BEHAVIOR
# ============================================================

engineered_df[
    "customer_amount_ratio"
] = (
    engineered_df["amount"]
    /
    engineered_df[
        "customer_previous_avg_amount"
    ].replace(
        0,
        np.nan
    )
)


engineered_df[
    "merchant_amount_ratio"
] = (
    engineered_df["amount"]
    /
    engineered_df[
        "merchant_previous_avg_amount"
    ].replace(
        0,
        np.nan
    )
)


engineered_df[
    "category_amount_ratio"
] = (
    engineered_df["amount"]
    /
    engineered_df[
        "category_previous_avg_amount"
    ].replace(
        0,
        np.nan
    )
)


# ============================================================
# 8. RENAME CUSTOMER / MERCHANT FEATURES
# ============================================================

engineered_df = engineered_df.rename(
    columns={

        # Customer
        "customer_previous_total_amount":
            "customer_previous_total_spend",

        "customer_previous_avg_amount":
            "customer_previous_avg_spend",

        "customer_previous_std_amount":
            "customer_previous_std_spend",

        "customer_previous_min_amount":
            "customer_previous_min_spend",

        "customer_previous_max_amount":
            "customer_previous_max_spend",

        "customer_amount_per_step":
            "customer_spend_per_step",

        # Merchant
        "merchant_previous_total_amount":
            "merchant_previous_total_received",

        "merchant_amount_per_step":
            "merchant_received_per_step",
    }
)


# ============================================================
# 9. REMOVE GLOBAL is_new_category
# ============================================================
#
# We agreed this adds little value.
#
# Customer-specific:
#
# is_new_category_for_customer
#
# will be created below.
#
# ============================================================

if "is_new_category" in engineered_df.columns:

    engineered_df = engineered_df.drop(
        columns=[
            "is_new_category"
        ]
    )


# ============================================================
# 10. RELATIONSHIP HISTORY FUNCTION
# ============================================================
#
# Used for:
#
#   customer ↔ merchant
#   customer ↔ category
#
# ============================================================


def build_relationship_history(
    source,
    relationship_cols,
    prefix
):

    temp = source[
        relationship_cols
        +
        [
            "step",
            "amount"
        ]
    ].copy()


    temp["_amount_squared"] = (
        temp["amount"] ** 2
    )


    # --------------------------------------------------------
    # Relationship activity by simulated day
    # --------------------------------------------------------

    step_stats = (
        temp
        .groupby(
            relationship_cols
            +
            [
                "step"
            ],
            observed=True,
            sort=True
        )
        .agg(

            step_transaction_count=(
                "amount",
                "size"
            ),

            step_total_amount=(
                "amount",
                "sum"
            ),

            step_amount_squared_sum=(
                "_amount_squared",
                "sum"
            ),

            step_min_amount=(
                "amount",
                "min"
            ),

            step_max_amount=(
                "amount",
                "max"
            )
        )
        .reset_index()
        .sort_values(
            relationship_cols
            +
            [
                "step"
            ]
        )
        .reset_index(drop=True)
    )


    grouped = step_stats.groupby(
        relationship_cols,
        sort=False
    )


    # ========================================================
    # PREVIOUS COUNT
    # ========================================================

    previous_count = (
        grouped[
            "step_transaction_count"
        ].cumsum()
        -
        step_stats[
            "step_transaction_count"
        ]
    )


    step_stats[
        f"{prefix}_previous_transaction_count"
    ] = previous_count


    # ========================================================
    # PREVIOUS TOTAL AMOUNT
    # ========================================================

    previous_total = (
        grouped[
            "step_total_amount"
        ].cumsum()
        -
        step_stats[
            "step_total_amount"
        ]
    )


    step_stats[
        f"{prefix}_previous_total_amount"
    ] = previous_total


    # ========================================================
    # PREVIOUS AVERAGE
    # ========================================================

    step_stats[
        f"{prefix}_previous_avg_amount"
    ] = (
        previous_total
        /
        previous_count.replace(
            0,
            np.nan
        )
    )


    # ========================================================
    # PREVIOUS STANDARD DEVIATION
    # ========================================================

    previous_squared_sum = (
        grouped[
            "step_amount_squared_sum"
        ].cumsum()
        -
        step_stats[
            "step_amount_squared_sum"
        ]
    )


    safe_count = previous_count.replace(
        0,
        np.nan
    )


    variance_numerator = (
        previous_squared_sum
        -
        (
            previous_total ** 2
            /
            safe_count
        )
    )


    variance_numerator = (
        variance_numerator
        .clip(lower=0)
    )


    step_stats[
        f"{prefix}_previous_std_amount"
    ] = np.sqrt(
        variance_numerator
        /
        (
            previous_count - 1
        ).where(
            previous_count > 1
        )
    )


    # ========================================================
    # PREVIOUS MIN
    # ========================================================

    step_stats["_shifted_min"] = (
        grouped[
            "step_min_amount"
        ].shift(1)
    )


    step_stats[
        f"{prefix}_previous_min_amount"
    ] = (
        step_stats
        .groupby(
            relationship_cols,
            sort=False
        )["_shifted_min"]
        .cummin()
    )


    # ========================================================
    # PREVIOUS MAX
    # ========================================================

    step_stats["_shifted_max"] = (
        grouped[
            "step_max_amount"
        ].shift(1)
    )


    step_stats[
        f"{prefix}_previous_max_amount"
    ] = (
        step_stats
        .groupby(
            relationship_cols,
            sort=False
        )["_shifted_max"]
        .cummax()
    )


    # ========================================================
    # TIME SINCE LAST RELATIONSHIP INTERACTION
    # ========================================================

    previous_step = grouped[
        "step"
    ].shift(1)


    step_stats[
        f"{prefix}_steps_since_last_transaction"
    ] = (
        step_stats["step"]
        -
        previous_step
    )


    # ========================================================
    # NEW RELATIONSHIP FLAG
    # ========================================================

    step_stats[
        f"is_new_{prefix}"
    ] = (
        previous_count == 0
    ).astype(int)


    # ========================================================
    # OUTPUT
    # ========================================================

    generated_columns = [
        col
        for col in step_stats.columns
        if (
            col.startswith(
                prefix + "_"
            )
            or
            col == f"is_new_{prefix}"
        )
    ]


    return step_stats[
        relationship_cols
        +
        [
            "step"
        ]
        +
        generated_columns
    ]


# ============================================================
# 11. CUSTOMER ↔ MERCHANT HISTORY
# ============================================================

customer_merchant_features = (
    build_relationship_history(
        source=source_df,
        relationship_cols=[
            "customer",
            "merchant"
        ],
        prefix="customer_merchant"
    )
)


# ============================================================
# 12. CUSTOMER ↔ CATEGORY HISTORY
# ============================================================

customer_category_features = (
    build_relationship_history(
        source=source_df,
        relationship_cols=[
            "customer",
            "category"
        ],
        prefix="customer_category"
    )
)


# ============================================================
# 13. MERGE RELATIONSHIP FEATURES
# ============================================================

engineered_df = (
    engineered_df
    .merge(
        customer_merchant_features,
        on=[
            "customer",
            "merchant",
            "step"
        ],
        how="left",
        validate="many_to_one"
    )
    .merge(
        customer_category_features,
        on=[
            "customer",
            "category",
            "step"
        ],
        how="left",
        validate="many_to_one"
    )
)


# ============================================================
# 14. CURRENT AMOUNT VS RELATIONSHIP NORMAL
# ============================================================

engineered_df[
    "customer_merchant_amount_ratio"
] = (
    engineered_df["amount"]
    /
    engineered_df[
        "customer_merchant_previous_avg_amount"
    ].replace(
        0,
        np.nan
    )
)


engineered_df[
    "customer_category_amount_ratio"
] = (
    engineered_df["amount"]
    /
    engineered_df[
        "customer_category_previous_avg_amount"
    ].replace(
        0,
        np.nan
    )
)


# ============================================================
# 15. CUSTOMER ↔ MERCHANT AFFINITY
# ============================================================

engineered_df[
    "customer_merchant_transaction_share"
] = (
    engineered_df[
        "customer_merchant_previous_transaction_count"
    ]
    /
    engineered_df[
        "customer_previous_transaction_count"
    ].replace(
        0,
        np.nan
    )
)


engineered_df[
    "customer_merchant_spend_share"
] = (
    engineered_df[
        "customer_merchant_previous_total_amount"
    ]
    /
    engineered_df[
        "customer_previous_total_spend"
    ].replace(
        0,
        np.nan
    )
)


# ============================================================
# 16. CUSTOMER ↔ CATEGORY AFFINITY
# ============================================================

engineered_df[
    "customer_category_transaction_share"
] = (
    engineered_df[
        "customer_category_previous_transaction_count"
    ]
    /
    engineered_df[
        "customer_previous_transaction_count"
    ].replace(
        0,
        np.nan
    )
)


engineered_df[
    "customer_category_spend_share"
] = (
    engineered_df[
        "customer_category_previous_total_amount"
    ]
    /
    engineered_df[
        "customer_previous_total_spend"
    ].replace(
        0,
        np.nan
    )
)


# ============================================================
# 17. CLEAN RELATIONSHIP FLAG NAMES
# ============================================================

engineered_df = engineered_df.rename(
    columns={

        "is_new_customer_merchant":
            "is_new_merchant_for_customer",

        "is_new_customer_category":
            "is_new_category_for_customer"
    }
)


# ============================================================
# 18. CHECK MERCHANT ↔ CATEGORY REDUNDANCY
# ============================================================

merchant_category_check = (
    source_df
    .groupby(
        "merchant"
    )["category"]
    .nunique()
)


max_categories_per_merchant = (
    merchant_category_check.max()
)


print("\n" + "=" * 70)
print("MERCHANT ↔ CATEGORY CHECK")
print("=" * 70)

print(
    "Maximum categories per merchant:",
    max_categories_per_merchant
)


if max_categories_per_merchant == 1:

    print(
        "Merchant ↔ Category relationship "
        "features are redundant and were NOT created."
    )

else:

    print(
        "Some merchants appear in multiple categories."
    )


# ============================================================
# 19. RESTORE ORIGINAL ROW ORDER
# ============================================================

engineered_df = (
    engineered_df
    .sort_values(
        "_row_id"
    )
    .reset_index(drop=True)
)


# ============================================================
# 20. VERIFY RAW COLUMNS DID NOT CHANGE
# ============================================================

print("\n" + "=" * 70)
print("RAW COLUMN ALIGNMENT")
print("=" * 70)


raw_check_columns = [
    "step",
    "customer",
    "merchant",
    "category",
    "amount",
    "fraud"
]


raw_alignment_results = {}


for col in raw_check_columns:

    left = (
        source_df[col]
        .reset_index(drop=True)
        .astype(str)
    )

    right = (
        engineered_df[col]
        .reset_index(drop=True)
        .astype(str)
    )


    identical_percentage = (
        (left == right).mean()
        * 100
    )


    raw_alignment_results[col] = (
        identical_percentage
    )


    print(
        f"{col:15} "
        f"{identical_percentage:.2f}% identical"
    )


# ============================================================
# 21. REMOVE INTERNAL ROW ID
# ============================================================

engineered_df = engineered_df.drop(
    columns=[
        "_row_id"
    ]
)


# ============================================================
# 22. STRUCTURAL AUDIT
# ============================================================

print("\n" + "=" * 70)
print("FEATURE AUDIT")
print("=" * 70)


print(
    f"Rows:    {engineered_df.shape[0]:,}"
)

print(
    f"Columns: {engineered_df.shape[1]:,}"
)


print(
    "Rows preserved:",
    len(engineered_df)
    ==
    len(source_df)
)


# ============================================================
# 23. MISSING VALUES
# ============================================================

missing_report = pd.DataFrame({

    "missing_count":
        engineered_df.isna().sum(),

    "missing_percentage":
        engineered_df.isna().mean()
        * 100
})


missing_report = (
    missing_report[
        missing_report[
            "missing_count"
        ] > 0
    ]
    .sort_values(
        "missing_percentage",
        ascending=False
    )
)


print(
    "\nColumns containing NaN:",
    len(missing_report)
)


display(
    missing_report
)


# ============================================================
# 24. INFINITE VALUES
# ============================================================

numeric_columns = (
    engineered_df
    .select_dtypes(
        include=np.number
    )
    .columns
)


inf_report = []


for col in numeric_columns:

    positive_inf = (
        np.isposinf(
            engineered_df[col]
        ).sum()
    )

    negative_inf = (
        np.isneginf(
            engineered_df[col]
        ).sum()
    )


    if (
        positive_inf > 0
        or
        negative_inf > 0
    ):

        inf_report.append({

            "feature": col,

            "positive_inf":
                positive_inf,

            "negative_inf":
                negative_inf,

            "total_inf":
                positive_inf
                +
                negative_inf
        })


inf_report = pd.DataFrame(
    inf_report
)


print("\nInfinite value check:")


if len(inf_report) == 0:

    print(
        "No infinite values found."
    )

else:

    display(
        inf_report
    )


# ============================================================
# 25. CONSTANT FEATURES
# ============================================================

constant_features = [
    col
    for col in engineered_df.columns
    if engineered_df[
        col
    ].nunique(
        dropna=False
    ) <= 1
]


print(
    "\nConstant features:",
    constant_features
)


# ============================================================
# 26. NEAR-CONSTANT FEATURES
# ============================================================

near_constant_rows = []


for col in engineered_df.columns:

    distribution = (
        engineered_df[col]
        .value_counts(
            normalize=True,
            dropna=False
        )
    )


    if len(distribution) == 0:
        continue


    dominant_percentage = (
        distribution.iloc[0]
        * 100
    )


    if dominant_percentage >= 99:

        near_constant_rows.append({

            "feature":
                col,

            "dominant_value":
                distribution.index[0],

            "dominant_percentage":
                dominant_percentage,

            "unique_values":
                engineered_df[
                    col
                ].nunique(
                    dropna=False
                )
        })


near_constant_report = (
    pd.DataFrame(
        near_constant_rows
    )
)


print(
    "\nNear-constant features:",
    len(near_constant_report)
)


if len(near_constant_report) > 0:

    display(
        near_constant_report
        .sort_values(
            "dominant_percentage",
            ascending=False
        )
    )


# ============================================================
# 27. SHARE VALIDATION
# ============================================================

share_columns = [
    col
    for col in engineered_df.columns
    if col.endswith(
        "_share"
    )
]


share_report = []


for col in share_columns:

    values = (
        engineered_df[col]
        .dropna()
    )


    share_report.append({

        "feature":
            col,

        "below_0":
            (values < 0).sum(),

        "above_1":
            (values > 1).sum(),

        "min":
            values.min(),

        "max":
            values.max()
    })


share_report = pd.DataFrame(
    share_report
)


print("\nShare validation:")


display(
    share_report
)


# ============================================================
# 28. HISTORICAL LEAKAGE VALIDATION
# ============================================================
#
# IMPORTANT:
#
# Uses source_df.
#
# clean_df -> source_df is the SAME source used to
# generate ALL features.
#
# ============================================================

print("\n" + "=" * 70)
print("HISTORICAL LEAKAGE VALIDATION")
print("=" * 70)


available_rows = (
    engineered_df[
        engineered_df["step"]
        >
        engineered_df["step"].min()
    ]
)


audit_rows = (
    available_rows
    .sample(
        n=min(
            20,
            len(available_rows)
        ),
        random_state=42
    )
)


leakage_results = []


for idx, row in audit_rows.iterrows():

    current_step = row["step"]

    customer = row["customer"]

    merchant = row["merchant"]

    category = row["category"]


    # --------------------------------------------------------
    # Customer
    # --------------------------------------------------------

    expected_customer = len(

        source_df[
            (
                source_df["customer"]
                ==
                customer
            )
            &
            (
                source_df["step"]
                <
                current_step
            )
        ]
    )


    # --------------------------------------------------------
    # Merchant
    # --------------------------------------------------------

    expected_merchant = len(

        source_df[
            (
                source_df["merchant"]
                ==
                merchant
            )
            &
            (
                source_df["step"]
                <
                current_step
            )
        ]
    )


    # --------------------------------------------------------
    # Category
    # --------------------------------------------------------

    expected_category = len(

        source_df[
            (
                source_df["category"]
                ==
                category
            )
            &
            (
                source_df["step"]
                <
                current_step
            )
        ]
    )


    # --------------------------------------------------------
    # Customer ↔ Merchant
    # --------------------------------------------------------

    expected_customer_merchant = len(

        source_df[
            (
                source_df["customer"]
                ==
                customer
            )
            &
            (
                source_df["merchant"]
                ==
                merchant
            )
            &
            (
                source_df["step"]
                <
                current_step
            )
        ]
    )


    # --------------------------------------------------------
    # Customer ↔ Category
    # --------------------------------------------------------

    expected_customer_category = len(

        source_df[
            (
                source_df["customer"]
                ==
                customer
            )
            &
            (
                source_df["category"]
                ==
                category
            )
            &
            (
                source_df["step"]
                <
                current_step
            )
        ]
    )


    leakage_results.append({

        "row":
            idx,

        "step":
            current_step,

        "customer_ok":
            expected_customer
            ==
            row[
                "customer_previous_transaction_count"
            ],

        "merchant_ok":
            expected_merchant
            ==
            row[
                "merchant_previous_transaction_count"
            ],

        "category_ok":
            expected_category
            ==
            row[
                "category_previous_transaction_count"
            ],

        "customer_merchant_ok":
            expected_customer_merchant
            ==
            row[
                "customer_merchant_previous_transaction_count"
            ],

        "customer_category_ok":
            expected_customer_category
            ==
            row[
                "customer_category_previous_transaction_count"
            ]
    })


leakage_results = pd.DataFrame(
    leakage_results
)


display(
    leakage_results
)


check_columns = [
    "customer_ok",
    "merchant_ok",
    "category_ok",
    "customer_merchant_ok",
    "customer_category_ok"
]


all_leakage_checks_passed = (
    leakage_results[
        check_columns
    ]
    .all()
    .all()
)


print(
    "\nHistorical leakage checks passed:",
    all_leakage_checks_passed
)


# ============================================================
# 29. SAME-STEP LEAKAGE CHECK
# ============================================================

print("\n" + "=" * 70)
print("SAME-STEP CHECK")
print("=" * 70)


same_step_groups = (

    source_df
    .groupby(
        [
            "customer",
            "step"
        ]
    )
    .size()
    .reset_index(
        name="same_step_count"
    )
)


same_step_groups = (
    same_step_groups[
        same_step_groups[
            "same_step_count"
        ] > 1
    ]
)


if len(same_step_groups) > 0:

    example = same_step_groups.iloc[0]

    example_customer = (
        example["customer"]
    )

    example_step = (
        example["step"]
    )


    expected_previous = len(

        source_df[
            (
                source_df["customer"]
                ==
                example_customer
            )
            &
            (
                source_df["step"]
                <
                example_step
            )
        ]
    )


    example_rows = (

        engineered_df[
            (
                engineered_df["customer"]
                ==
                example_customer
            )
            &
            (
                engineered_df["step"]
                ==
                example_step
            )
        ]
    )


    print(
        "Customer:",
        example_customer
    )

    print(
        "Step:",
        example_step
    )

    print(
        "Transactions occurring in same step:",
        len(example_rows)
    )

    print(
        "Expected historical count:",
        expected_previous
    )


    display(

        example_rows[
            [
                "step",
                "customer",
                "amount",
                "customer_previous_transaction_count"
            ]
        ]
    )


else:

    print(
        "No customer has multiple transactions "
        "within the same step."
    )


# ============================================================
# 30. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PIPELINE COMPLETE")
print("=" * 70)


print(
    f"Source rows:      "
    f"{len(source_df):,}"
)

print(
    f"Engineered rows:  "
    f"{len(engineered_df):,}"
)

print(
    f"Engineered columns: "
    f"{engineered_df.shape[1]}"
)

print(
    f"Columns with NaN: "
    f"{len(missing_report)}"
)

print(
    f"Constant features: "
    f"{len(constant_features)}"
)

print(
    f"Near-constant features: "
    f"{len(near_constant_report)}"
)

print(
    "Historical leakage validation:",
    all_leakage_checks_passed
)


print(
    "\nFinal dataframe name: engineered_df"
)
```

```python
# ============================================================
# STRUCTURAL MISSING-VALUE HANDLING
# ============================================================
#
# INPUT:
#     engineered_df
#
# PURPOSE:
#     Handle NaN values created naturally by historical
#     feature engineering.
#
# IMPORTANT:
#     We first handle NaNs according to their BUSINESS MEANING.
#
#     Only after that do we apply the selected fallback:
#
#         "zero" -> remaining NaNs become 0
#         "drop" -> rows containing remaining NaNs are removed
#
# ============================================================

import numpy as np
import pandas as pd


# ============================================================
# CONFIGURATION
# ============================================================

MISSING_STRATEGY = "zero"

# Options:
#
# "zero"
#     Replace any remaining engineered NaN values with 0.
#
# "drop"
#     Drop rows that still contain NaN values.
#
# Recommended for this project:
#     "zero"
# ============================================================


if MISSING_STRATEGY not in {"zero", "drop"}:
    raise ValueError(
        "MISSING_STRATEGY must be either 'zero' or 'drop'"
    )


print("=" * 70)
print("STRUCTURAL MISSING-VALUE HANDLING")
print("=" * 70)

print(
    "Initial missing values:",
    engineered_df.isna().sum().sum()
)


# ============================================================
# 1. NEW CUSTOMER
# ============================================================
#
# If:
#
#     is_new_customer = 1
#
# then the customer has no historical behavior.
#
# Historical customer features therefore become 0.
#
# ============================================================

customer_no_history_cols = [

    "customer_previous_transaction_count",

    "customer_previous_total_spend",
    "customer_previous_avg_spend",
    "customer_previous_std_spend",
    "customer_previous_min_spend",
    "customer_previous_max_spend",

    "customer_transactions_per_step",
    "customer_spend_per_step",

    "customer_steps_since_last_transaction",

    "customer_transactions_last_1_steps",
    "customer_amount_last_1_steps",

    "customer_transactions_last_3_steps",
    "customer_amount_last_3_steps",

    "customer_transactions_last_6_steps",
    "customer_amount_last_6_steps",

    "customer_transactions_last_24_steps",
    "customer_amount_last_24_steps",

    "customer_velocity_ratio",
    "customer_recent_amount_ratio",

    "customer_amount_ratio"
]


new_customer_mask = (
    engineered_df["is_new_customer"] == 1
)


engineered_df.loc[
    new_customer_mask,
    customer_no_history_cols
] = 0


# ============================================================
# 2. NEW MERCHANT
# ============================================================

merchant_no_history_cols = [

    "merchant_previous_transaction_count",

    "merchant_previous_total_received",
    "merchant_previous_avg_amount",
    "merchant_previous_std_amount",
    "merchant_previous_min_amount",
    "merchant_previous_max_amount",

    "merchant_transactions_per_step",
    "merchant_received_per_step",

    "merchant_steps_since_last_transaction",

    "merchant_transactions_last_1_steps",
    "merchant_amount_last_1_steps",

    "merchant_transactions_last_3_steps",
    "merchant_amount_last_3_steps",

    "merchant_transactions_last_6_steps",
    "merchant_amount_last_6_steps",

    "merchant_transactions_last_24_steps",
    "merchant_amount_last_24_steps",

    "merchant_velocity_ratio",
    "merchant_recent_amount_ratio",

    "merchant_amount_ratio"
]


new_merchant_mask = (
    engineered_df["is_new_merchant"] == 1
)


engineered_df.loc[
    new_merchant_mask,
    merchant_no_history_cols
] = 0


# ============================================================
# 3. NEW CUSTOMER ↔ MERCHANT RELATIONSHIP
# ============================================================
#
# If:
#
#     is_new_merchant_for_customer = 1
#
# this customer has NEVER used this merchant before.
#
# ============================================================

customer_merchant_no_history_cols = [

    "customer_merchant_previous_transaction_count",

    "customer_merchant_previous_total_amount",
    "customer_merchant_previous_avg_amount",
    "customer_merchant_previous_std_amount",
    "customer_merchant_previous_min_amount",
    "customer_merchant_previous_max_amount",

    "customer_merchant_steps_since_last_transaction",

    "customer_merchant_amount_ratio",

    "customer_merchant_transaction_share",
    "customer_merchant_spend_share"
]


new_customer_merchant_mask = (
    engineered_df[
        "is_new_merchant_for_customer"
    ] == 1
)


engineered_df.loc[
    new_customer_merchant_mask,
    customer_merchant_no_history_cols
] = 0


# ============================================================
# 4. NEW CUSTOMER ↔ CATEGORY RELATIONSHIP
# ============================================================

customer_category_no_history_cols = [

    "customer_category_previous_transaction_count",

    "customer_category_previous_total_amount",
    "customer_category_previous_avg_amount",
    "customer_category_previous_std_amount",
    "customer_category_previous_min_amount",
    "customer_category_previous_max_amount",

    "customer_category_steps_since_last_transaction",

    "customer_category_amount_ratio",

    "customer_category_transaction_share",
    "customer_category_spend_share"
]


new_customer_category_mask = (
    engineered_df[
        "is_new_category_for_customer"
    ] == 1
)


engineered_df.loc[
    new_customer_category_mask,
    customer_category_no_history_cols
] = 0


# ============================================================
# 5. CATEGORY WITH NO PREVIOUS HISTORY
# ============================================================
#
# We intentionally removed is_new_category.
#
# Instead:
#
# category_previous_transaction_count == 0
#
# tells us there was no category history.
#
# ============================================================

category_no_history_cols = [

    "category_previous_avg_amount",
    "category_previous_std_amount",
    "category_previous_min_amount",
    "category_previous_max_amount",

    "category_steps_since_last_transaction",

    "category_transactions_per_step",
    "category_amount_per_step",

    "category_velocity_ratio",
    "category_recent_amount_ratio",

    "category_amount_ratio"
]


category_no_history_mask = (
    engineered_df[
        "category_previous_transaction_count"
    ] == 0
)


engineered_df.loc[
    category_no_history_mask,
    category_no_history_cols
] = 0


# ============================================================
# 6. STANDARD DEVIATION WITH ONLY ONE HISTORICAL OBSERVATION
# ============================================================
#
# Sample standard deviation requires at least 2 observations.
#
# When previous_count <= 1:
#
#     observed historical variation = 0
#
# ============================================================


# Customer
engineered_df.loc[
    engineered_df[
        "customer_previous_transaction_count"
    ] <= 1,
    "customer_previous_std_spend"
] = 0


# Merchant
engineered_df.loc[
    engineered_df[
        "merchant_previous_transaction_count"
    ] <= 1,
    "merchant_previous_std_amount"
] = 0


# Category
engineered_df.loc[
    engineered_df[
        "category_previous_transaction_count"
    ] <= 1,
    "category_previous_std_amount"
] = 0


# Customer ↔ Merchant
engineered_df.loc[
    engineered_df[
        "customer_merchant_previous_transaction_count"
    ] <= 1,
    "customer_merchant_previous_std_amount"
] = 0


# Customer ↔ Category
engineered_df.loc[
    engineered_df[
        "customer_category_previous_transaction_count"
    ] <= 1,
    "customer_category_previous_std_amount"
] = 0


# ============================================================
# 7. CHECK WHAT REMAINS
# ============================================================

remaining_missing = (
    engineered_df
    .isna()
    .sum()
)

remaining_missing = (
    remaining_missing[
        remaining_missing > 0
    ]
    .sort_values(
        ascending=False
    )
)


print(
    "\nMissing after structural handling:",
    engineered_df.isna().sum().sum()
)


if len(remaining_missing) > 0:

    display(
        remaining_missing.to_frame(
            "remaining_missing"
        )
    )

else:

    print(
        "No remaining missing values "
        "after structural handling."
    )


# ============================================================
# 8. FINAL FALLBACK STRATEGY
# ============================================================

rows_before = len(engineered_df)


if engineered_df.isna().sum().sum() > 0:

    # --------------------------------------------------------
    # OPTION A — ZERO
    # --------------------------------------------------------

    if MISSING_STRATEGY == "zero":

        remaining_cols = (
            engineered_df
            .columns[
                engineered_df
                .isna()
                .any()
            ]
            .tolist()
        )


        # Verify remaining missing features are numerical
        non_numeric_missing_cols = [

            col
            for col in remaining_cols

            if not pd.api.types.is_numeric_dtype(
                engineered_df[col]
            )
        ]


        if non_numeric_missing_cols:

            raise ValueError(
                "Remaining NaN values exist in "
                "non-numeric columns: "
                f"{non_numeric_missing_cols}"
            )


        engineered_df[
            remaining_cols
        ] = (
            engineered_df[
                remaining_cols
            ]
            .fillna(0)
        )


        print(
            "\nFallback applied:"
            "\nRemaining numeric NaNs → 0"
        )


    # --------------------------------------------------------
    # OPTION B — DROP ROWS
    # --------------------------------------------------------

    elif MISSING_STRATEGY == "drop":

        rows_with_missing = (
            engineered_df
            .isna()
            .any(axis=1)
            .sum()
        )


        engineered_df = (
            engineered_df
            .dropna()
            .reset_index(drop=True)
        )


        print(
            f"\nFallback applied:"
            f"\nDropped {rows_with_missing:,} rows "
            f"containing remaining NaN values."
        )


# ============================================================
# 9. FINAL VALIDATION
# ============================================================

rows_after = len(engineered_df)

remaining_nan_count = (
    engineered_df
    .isna()
    .sum()
    .sum()
)


print("\n" + "=" * 70)
print("FINAL MISSING-VALUE SUMMARY")
print("=" * 70)

print(
    f"Strategy: {MISSING_STRATEGY}"
)

print(
    f"Rows before: {rows_before:,}"
)

print(
    f"Rows after:  {rows_after:,}"
)

print(
    f"Rows removed: "
    f"{rows_before - rows_after:,}"
)

print(
    f"Remaining NaN values: "
    f"{remaining_nan_count:,}"
)


if remaining_nan_count == 0:

    print(
        "\nPASS: engineered_df contains "
        "no missing values."
    )

else:

    print(
        "\nWARNING: missing values still remain."
    )
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    RobustScaler,
    OneHotEncoder,
    OrdinalEncoder
)


# ============================================================
# CONFIG
# ============================================================

PREPROCESS_CONFIG = {

    # NUMERIC
    "numeric_enabled": True,

    # "standard"
    # "minmax"
    # "robust"
    # "none"
    "numeric_method": "standard",


    # CATEGORICAL
    "categorical_enabled": True,

    # "onehot"
    # "ordinal"
    # "none"
    "categorical_method": "onehot"
}


# ============================================================
# FEATURE TYPES
# ============================================================

categorical_features = [
    "age",
    "gender",
    "category"
]


numeric_features = [
    col
    for col in X_train.columns
    if col not in categorical_features
]


# ============================================================
# NUMERIC TRANSFORMER
# ============================================================

if PREPROCESS_CONFIG["numeric_enabled"]:

    method = PREPROCESS_CONFIG["numeric_method"]

    if method == "standard":
        numeric_transformer = StandardScaler()

    elif method == "minmax":
        numeric_transformer = MinMaxScaler()

    elif method == "robust":
        numeric_transformer = RobustScaler()

    elif method == "none":
        numeric_transformer = "passthrough"

    else:
        raise ValueError(
            f"Unknown numeric method: {method}"
        )

else:

    numeric_transformer = "passthrough"


# ============================================================
# CATEGORICAL TRANSFORMER
# ============================================================

if PREPROCESS_CONFIG["categorical_enabled"]:

    method = PREPROCESS_CONFIG["categorical_method"]

    if method == "onehot":

        categorical_transformer = OneHotEncoder(
            handle_unknown="ignore"
        )

    elif method == "ordinal":

        categorical_transformer = OrdinalEncoder(
            handle_unknown="use_encoded_value",
            unknown_value=-1
        )

    elif method == "none":

        categorical_transformer = "passthrough"

    else:

        raise ValueError(
            f"Unknown categorical method: {method}"
        )

else:

    categorical_transformer = "passthrough"


# ============================================================
# BUILD PREPROCESSOR
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[

        (
            "numeric",
            numeric_transformer,
            numeric_features
        ),

        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ]
)
```

```python
def evaluate_model(
    model,
    Xtr,
    Xv,
    ytr,
    yv,
    model_name,
    feature_set_name,
    threshold=0.50
):

    model.fit(
        Xtr,
        ytr
    )

    # Train probabilities
    train_prob = model.predict_proba(
        Xtr
    )[:, 1]

    # Validation probabilities
    val_prob = model.predict_proba(
        Xv
    )[:, 1]


    train_pred = (
        train_prob >= threshold
    ).astype(int)

    val_pred = (
        val_prob >= threshold
    ).astype(int)


    return {

        "feature_set":
            feature_set_name,

        "model":
            model_name,

        "features":
            Xtr.shape[1],

        # Train
        "train_precision":
            precision_score(
                ytr,
                train_pred,
                zero_division=0
            ),

        "train_recall":
            recall_score(
                ytr,
                train_pred,
                zero_division=0
            ),

        "train_f1":
            f1_score(
                ytr,
                train_pred,
                zero_division=0
            ),

        # Validation
        "val_precision":
            precision_score(
                yv,
                val_pred,
                zero_division=0
            ),

        "val_recall":
            recall_score(
                yv,
                val_pred,
                zero_division=0
            ),

        "val_f1":
            f1_score(
                yv,
                val_pred,
                zero_division=0
            ),

        "val_pr_auc":
            average_precision_score(
                yv,
                val_prob
            ),

        "val_roc_auc":
            roc_auc_score(
                yv,
                val_prob
            )
    }
```

```python
# ============================================================
# MODEL TRAINING + THRESHOLD OPTIMIZATION
# ============================================================
#
# Expected existing variables:
#
# y_train
# y_val
#
# X_train_all
# X_val_all
#
# X_train_rf
# X_val_rf
#
# Logistic Regression:
#     uses ALL processed features
#
# Random Forest:
#     uses RF-selected features
#
# TEST DATA IS NOT USED HERE.
# ============================================================


import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    fbeta_score,
    average_precision_score,
    roc_auc_score,
    confusion_matrix,
    precision_recall_curve
)

from scipy.optimize import minimize_scalar


# ============================================================
# 1. CONFIGURATION
# ============================================================

THRESHOLD_CONFIG = {

    # Default threshold
    "default_threshold": 0.50,

    # Grid search
    "grid_min": 0.01,
    "grid_max": 0.99,
    "grid_step": 0.01,

    # Objective:
    #
    # "f1" -> equal importance precision / recall
    #
    # "f2" -> recall is more important
    #
    "objective": "f1"
}


# ============================================================
# 2. BUILD MODELS
# ============================================================

logistic_model = LogisticRegression(
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)


random_forest_model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)


# ============================================================
# 3. TRAIN MODELS
# ============================================================

print("=" * 70)
print("TRAINING MODELS")
print("=" * 70)


# Logistic Regression
# Uses all processed features
logistic_model.fit(
    X_train_all,
    y_train
)

print("Logistic Regression trained.")


# Random Forest
# Uses RF-selected feature set
random_forest_model.fit(
    X_train_rf,
    y_train
)

print("Random Forest trained.")


# ============================================================
# 4. SAVE TRAIN + VALIDATION PROBABILITIES
# ============================================================

probabilities = {

    "Logistic Regression": {

        "train":
            logistic_model.predict_proba(
                X_train_all
            )[:, 1],

        "validation":
            logistic_model.predict_proba(
                X_val_all
            )[:, 1]
    },


    "Random Forest": {

        "train":
            random_forest_model.predict_proba(
                X_train_rf
            )[:, 1],

        "validation":
            random_forest_model.predict_proba(
                X_val_rf
            )[:, 1]
    }
}


print("\nProbabilities saved.")


# ============================================================
# 5. METRIC FUNCTION
# ============================================================

def calculate_threshold_metrics(
    y_true,
    probability,
    threshold
):

    prediction = (
        probability >= threshold
    ).astype(int)


    precision = precision_score(
        y_true,
        prediction,
        zero_division=0
    )


    recall = recall_score(
        y_true,
        prediction,
        zero_division=0
    )


    f1 = f1_score(
        y_true,
        prediction,
        zero_division=0
    )


    f2 = fbeta_score(
        y_true,
        prediction,
        beta=2,
        zero_division=0
    )


    tn, fp, fn, tp = confusion_matrix(
        y_true,
        prediction,
        labels=[0, 1]
    ).ravel()


    return {

        "threshold":
            float(threshold),

        "precision":
            precision,

        "recall":
            recall,

        "f1":
            f1,

        "f2":
            f2,

        "tn":
            int(tn),

        "fp":
            int(fp),

        "fn":
            int(fn),

        "tp":
            int(tp)
    }


# ============================================================
# 6. DEFAULT THRESHOLD = 0.50
# ============================================================

default_results = []


for model_name in probabilities:

    train_prob = (
        probabilities[
            model_name
        ]["train"]
    )

    val_prob = (
        probabilities[
            model_name
        ]["validation"]
    )


    train_metrics = calculate_threshold_metrics(
        y_train,
        train_prob,
        THRESHOLD_CONFIG[
            "default_threshold"
        ]
    )


    val_metrics = calculate_threshold_metrics(
        y_val,
        val_prob,
        THRESHOLD_CONFIG[
            "default_threshold"
        ]
    )


    default_results.append({

        "model":
            model_name,

        "method":
            "Default 0.50",

        "threshold":
            THRESHOLD_CONFIG[
                "default_threshold"
            ],

        "train_precision":
            train_metrics["precision"],

        "train_recall":
            train_metrics["recall"],

        "train_f1":
            train_metrics["f1"],

        "val_precision":
            val_metrics["precision"],

        "val_recall":
            val_metrics["recall"],

        "val_f1":
            val_metrics["f1"],

        "val_f2":
            val_metrics["f2"],

        "tp":
            val_metrics["tp"],

        "fp":
            val_metrics["fp"],

        "fn":
            val_metrics["fn"],

        "tn":
            val_metrics["tn"],

        "pr_auc":
            average_precision_score(
                y_val,
                val_prob
            ),

        "roc_auc":
            roc_auc_score(
                y_val,
                val_prob
            )
    })


default_results_df = pd.DataFrame(
    default_results
)


print("\n" + "=" * 70)
print("DEFAULT THRESHOLD RESULTS")
print("=" * 70)

display(
    default_results_df
)


# ============================================================
# 7. GRID SEARCH
# ============================================================

def grid_threshold_search(
    y_true,
    probability
):

    thresholds = np.arange(

        THRESHOLD_CONFIG[
            "grid_min"
        ],

        THRESHOLD_CONFIG[
            "grid_max"
        ] + THRESHOLD_CONFIG[
            "grid_step"
        ],

        THRESHOLD_CONFIG[
            "grid_step"
        ]
    )


    rows = []


    for threshold in thresholds:

        metrics = calculate_threshold_metrics(
            y_true,
            probability,
            threshold
        )

        rows.append(
            metrics
        )


    results = pd.DataFrame(
        rows
    )


    objective = (
        THRESHOLD_CONFIG[
            "objective"
        ]
    )


    best_row = (

        results
        .sort_values(
            objective,
            ascending=False
        )
        .iloc[0]
    )


    return results, best_row


# ============================================================
# 8. SCIPY OPTIMIZATION
# ============================================================

def scipy_threshold_search(
    y_true,
    probability
):

    objective_name = (
        THRESHOLD_CONFIG[
            "objective"
        ]
    )


    def objective(
        threshold
    ):

        metrics = calculate_threshold_metrics(
            y_true,
            probability,
            threshold
        )

        # scipy minimizes
        return -metrics[
            objective_name
        ]


    result = minimize_scalar(

        objective,

        bounds=(
            0.001,
            0.999
        ),

        method="bounded",

        options={
            "xatol": 0.0001
        }
    )


    best_threshold = (
        result.x
    )


    best_metrics = calculate_threshold_metrics(
        y_true,
        probability,
        best_threshold
    )


    return result, best_metrics


# ============================================================
# 9. EXACT PRECISION-RECALL THRESHOLD SEARCH
# ============================================================
#
# This checks thresholds based on actual predicted
# probability values.
#
# ============================================================

def precision_recall_threshold_search(
    y_true,
    probability
):

    precision, recall, thresholds = (
        precision_recall_curve(
            y_true,
            probability
        )
    )


    # Last precision / recall pair has no threshold
    precision = precision[:-1]
    recall = recall[:-1]


    # F1
    f1 = (
        2
        * precision
        * recall
        /
        (
            precision
            + recall
            + 1e-12
        )
    )


    # F2
    beta = 2

    f2 = (
        (1 + beta**2)
        * precision
        * recall
        /
        (
            beta**2
            * precision
            + recall
            + 1e-12
        )
    )


    if (
        THRESHOLD_CONFIG[
            "objective"
        ] == "f1"
    ):

        scores = f1

    else:

        scores = f2


    best_index = np.argmax(
        scores
    )


    best_threshold = (
        thresholds[
            best_index
        ]
    )


    best_metrics = calculate_threshold_metrics(
        y_true,
        probability,
        best_threshold
    )


    return best_metrics


# ============================================================
# 10. RUN ALL THRESHOLD METHODS
# ============================================================

threshold_summary = []

grid_results_by_model = {}


for model_name in probabilities:

    print(
        "\n",
        "=" * 70
    )

    print(
        model_name.upper()
    )

    print(
        "=" * 70
    )


    val_probability = (
        probabilities[
            model_name
        ]["validation"]
    )


    # --------------------------------------------------------
    # DEFAULT
    # --------------------------------------------------------

    default_metrics = (
        calculate_threshold_metrics(

            y_val,

            val_probability,

            THRESHOLD_CONFIG[
                "default_threshold"
            ]
        )
    )


    threshold_summary.append({

        "model":
            model_name,

        "method":
            "Default",

        **default_metrics
    })


    # --------------------------------------------------------
    # GRID SEARCH
    # --------------------------------------------------------

    grid_results, grid_best = (
        grid_threshold_search(

            y_val,

            val_probability
        )
    )


    grid_results_by_model[
        model_name
    ] = grid_results


    threshold_summary.append({

        "model":
            model_name,

        "method":
            "Grid Search",

        **grid_best.to_dict()
    })


    # --------------------------------------------------------
    # SCIPY
    # --------------------------------------------------------

    scipy_result, scipy_best = (
        scipy_threshold_search(

            y_val,

            val_probability
        )
    )


    threshold_summary.append({

        "model":
            model_name,

        "method":
            "SciPy",

        **scipy_best
    })


    # --------------------------------------------------------
    # PRECISION-RECALL CURVE
    # --------------------------------------------------------

    pr_best = (
        precision_recall_threshold_search(

            y_val,

            val_probability
        )
    )


    threshold_summary.append({

        "model":
            model_name,

        "method":
            "Precision-Recall Curve",

        **pr_best
    })


# ============================================================
# 11. FINAL COMPARISON
# ============================================================

threshold_results_df = (
    pd.DataFrame(
        threshold_summary
    )
)


objective = (
    THRESHOLD_CONFIG[
        "objective"
    ]
)


threshold_results_df = (

    threshold_results_df
    .sort_values(

        objective,

        ascending=False
    )
    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 80)
print("THRESHOLD OPTIMIZATION RESULTS")
print("=" * 80)


display(

    threshold_results_df[
        [
            "model",
            "method",
            "threshold",
            "precision",
            "recall",
            "f1",
            "f2",
            "tp",
            "fp",
            "fn",
            "tn"
        ]
    ]
)


# ============================================================
# 12. BEST RESULT PER MODEL
# ============================================================

best_per_model = (

    threshold_results_df
    .sort_values(
        objective,
        ascending=False
    )
    .groupby(
        "model",
        as_index=False
    )
    .first()
)


print("\n" + "=" * 80)
print(
    f"BEST THRESHOLD PER MODEL "
    f"USING {objective.upper()}"
)
print("=" * 80)


display(

    best_per_model[
        [
            "model",
            "method",
            "threshold",
            "precision",
            "recall",
            "f1",
            "f2",
            "tp",
            "fp",
            "fn"
        ]
    ]
)


# ============================================================
# 13. OPTIONAL BUSINESS-CONSTRAINED SEARCH
# ============================================================
#
# Example:
#
# Find the threshold with the highest precision
# while keeping recall >= 90%.
#
# ============================================================

MIN_RECALL = 0.90


business_results = []


for model_name in probabilities:

    grid_df = (
        grid_results_by_model[
            model_name
        ]
    )


    valid = grid_df[
        grid_df["recall"]
        >= MIN_RECALL
    ]


    if len(valid) > 0:

        best = (

            valid
            .sort_values(
                "precision",
                ascending=False
            )
            .iloc[0]
        )


        business_results.append({

            "model":
                model_name,

            "minimum_recall":
                MIN_RECALL,

            **best.to_dict()
        })


business_results_df = (
    pd.DataFrame(
        business_results
    )
)


print("\n" + "=" * 80)
print(
    f"HIGHEST PRECISION WITH "
    f"RECALL >= {MIN_RECALL:.0%}"
)
print("=" * 80)


if len(
    business_results_df
) > 0:

    display(
        business_results_df[
            [
                "model",
                "threshold",
                "precision",
                "recall",
                "f1",
                "tp",
                "fp",
                "fn"
            ]
        ]
    )

else:

    print(
        "No threshold satisfies "
        "the recall constraint."
    )


# ============================================================
# 14. DONE
# ============================================================

print("\n" + "=" * 80)
print("THRESHOLD ANALYSIS COMPLETE")
print("=" * 80)

print(
    "\nTest set has NOT been used."
)

print(
    "Do not evaluate test data until "
    "the final model and threshold are selected."
)
```

```python
# ============================================================
# RANDOM FOREST TUNING
# Goal:
#   - Strong validation performance
#   - Reduce train -> validation overfitting gap
#
# TEST SET IS NOT USED
# ============================================================

import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import ParameterSampler

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    roc_auc_score,
    confusion_matrix,
    precision_recall_curve
)


# ============================================================
# 1. CONFIGURATION
# ============================================================

TUNING_CONFIG = {

    # Number of parameter combinations to test
    "n_iterations": 25,

    # Initial threshold during model comparison
    "comparison_threshold": 0.50,

    # Desired maximum Train F1 - Validation F1 gap
    "max_f1_gap": 0.08,

    # How strongly to punish overfitting
    "gap_penalty": 0.50,

    "random_state": 42
}


# ============================================================
# 2. PARAMETER SEARCH SPACE
# ============================================================

param_space = {

    # Number of trees
    "n_estimators": [
        100,
        150,
        200,
        300
    ],

    # Major control for overfitting
    "max_depth": [
        6,
        8,
        10,
        12,
        15,
        20
    ],

    # Larger values reduce overfitting
    "min_samples_split": [
        2,
        5,
        10,
        20,
        50
    ],

    # Very useful for reducing memorization
    "min_samples_leaf": [
        1,
        2,
        5,
        10,
        20
    ],

    # Number of features considered per split
    "max_features": [
        "sqrt",
        "log2",
        0.5,
        0.75
    ],

    # Sampling observations also helps regularization
    "max_samples": [
        0.60,
        0.75,
        0.90,
        None
    ]
}


# ============================================================
# 3. METRIC FUNCTION
# ============================================================

def evaluate_probabilities(
    y_true,
    probabilities,
    threshold=0.50
):

    predictions = (
        probabilities >= threshold
    ).astype(int)

    return {

        "precision":
            precision_score(
                y_true,
                predictions,
                zero_division=0
            ),

        "recall":
            recall_score(
                y_true,
                predictions,
                zero_division=0
            ),

        "f1":
            f1_score(
                y_true,
                predictions,
                zero_division=0
            ),

        "pr_auc":
            average_precision_score(
                y_true,
                probabilities
            ),

        "roc_auc":
            roc_auc_score(
                y_true,
                probabilities
            )
    }


# ============================================================
# 4. GENERATE RANDOM PARAMETER COMBINATIONS
# ============================================================

parameter_combinations = list(
    ParameterSampler(
        param_space,
        n_iter=TUNING_CONFIG["n_iterations"],
        random_state=TUNING_CONFIG["random_state"]
    )
)


print("=" * 80)
print("RANDOM FOREST HYPERPARAMETER TUNING")
print("=" * 80)

print(
    "Configurations to test:",
    len(parameter_combinations)
)


# ============================================================
# 5. TRAIN / VALIDATION SEARCH
# ============================================================

tuning_results = []

trained_models = {}


for i, params in enumerate(
    parameter_combinations,
    start=1
):

    print(
        f"\n[{i}/{len(parameter_combinations)}]",
        params
    )


    # --------------------------------------------------------
    # Build model
    # --------------------------------------------------------

    model = RandomForestClassifier(

        **params,

        class_weight="balanced",

        bootstrap=True,

        random_state=42,

        n_jobs=-1
    )


    # --------------------------------------------------------
    # Train
    # --------------------------------------------------------

    model.fit(
        X_train_rf,
        y_train
    )


    # --------------------------------------------------------
    # Probabilities
    # --------------------------------------------------------

    train_prob = model.predict_proba(
        X_train_rf
    )[:, 1]


    val_prob = model.predict_proba(
        X_val_rf
    )[:, 1]


    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    train_metrics = evaluate_probabilities(
        y_train,
        train_prob,
        TUNING_CONFIG["comparison_threshold"]
    )


    val_metrics = evaluate_probabilities(
        y_val,
        val_prob,
        TUNING_CONFIG["comparison_threshold"]
    )


    # --------------------------------------------------------
    # Generalization gaps
    # --------------------------------------------------------

    f1_gap = (
        train_metrics["f1"]
        -
        val_metrics["f1"]
    )


    pr_auc_gap = (
        train_metrics["pr_auc"]
        -
        val_metrics["pr_auc"]
    )


    # --------------------------------------------------------
    # Combined selection score
    #
    # Reward validation PR-AUC.
    # Penalize excessive train-validation F1 difference.
    # --------------------------------------------------------

    selection_score = (

        val_metrics["pr_auc"]

        -

        TUNING_CONFIG["gap_penalty"]
        *
        max(f1_gap, 0)
    )


    tuning_results.append({

        "iteration": i,

        **params,

        # Train
        "train_precision":
            train_metrics["precision"],

        "train_recall":
            train_metrics["recall"],

        "train_f1":
            train_metrics["f1"],

        "train_pr_auc":
            train_metrics["pr_auc"],


        # Validation
        "val_precision":
            val_metrics["precision"],

        "val_recall":
            val_metrics["recall"],

        "val_f1":
            val_metrics["f1"],

        "val_pr_auc":
            val_metrics["pr_auc"],

        "val_roc_auc":
            val_metrics["roc_auc"],


        # Generalization
        "f1_gap":
            f1_gap,

        "pr_auc_gap":
            pr_auc_gap,

        "selection_score":
            selection_score
    })


    # Keep trained model
    trained_models[i] = model


# ============================================================
# 6. RESULTS
# ============================================================

tuning_results_df = pd.DataFrame(
    tuning_results
)


tuning_results_df = (
    tuning_results_df
    .sort_values(
        "selection_score",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 80)
print("TOP RANDOM FOREST CONFIGURATIONS")
print("=" * 80)


display(

    tuning_results_df[
        [
            "iteration",

            "n_estimators",
            "max_depth",
            "min_samples_split",
            "min_samples_leaf",
            "max_features",
            "max_samples",

            "train_f1",
            "val_f1",
            "f1_gap",

            "train_pr_auc",
            "val_pr_auc",
            "pr_auc_gap",

            "val_precision",
            "val_recall",

            "selection_score"
        ]
    ].head(15)

)


# ============================================================
# 7. FIND MODELS WITH ACCEPTABLE GENERALIZATION GAP
# ============================================================

acceptable_models = tuning_results_df[

    tuning_results_df["f1_gap"]
    <= TUNING_CONFIG["max_f1_gap"]

].copy()


print("\n" + "=" * 80)
print("GENERALIZATION CHECK")
print("=" * 80)


if len(acceptable_models) > 0:

    print(
        f"{len(acceptable_models)} configurations "
        f"have F1 gap <= "
        f"{TUNING_CONFIG['max_f1_gap']:.2f}"
    )


    # Among sufficiently regularized models,
    # prioritize validation PR-AUC

    best_row = (

        acceptable_models
        .sort_values(
            [
                "val_pr_auc",
                "val_f1"
            ],
            ascending=False
        )
        .iloc[0]
    )


else:

    print(
        "No configuration reached the desired gap."
    )

    print(
        "Using best combined generalization score instead."
    )


    best_row = (
        tuning_results_df
        .iloc[0]
    )


# ============================================================
# 8. GET BEST TRAINED MODEL
# ============================================================

best_iteration = int(
    best_row["iteration"]
)


best_rf_model = trained_models[
    best_iteration
]


print("\n" + "=" * 80)
print("SELECTED RANDOM FOREST")
print("=" * 80)


display(
    best_row[
        [
            "n_estimators",
            "max_depth",
            "min_samples_split",
            "min_samples_leaf",
            "max_features",
            "max_samples",

            "train_precision",
            "train_recall",
            "train_f1",

            "val_precision",
            "val_recall",
            "val_f1",

            "val_pr_auc",
            "val_roc_auc",

            "f1_gap",
            "pr_auc_gap"
        ]
    ]
)


# ============================================================
# 9. GET VALIDATION PROBABILITIES FROM BEST MODEL
# ============================================================

best_val_prob = best_rf_model.predict_proba(
    X_val_rf
)[:, 1]


best_train_prob = best_rf_model.predict_proba(
    X_train_rf
)[:, 1]


# ============================================================
# 10. RE-OPTIMIZE THRESHOLD
# ============================================================
#
# Changing RF parameters changes the probability distribution,
# therefore the previous threshold 0.26 is NOT automatically
# valid anymore.
#
# ============================================================

precision, recall, thresholds = (
    precision_recall_curve(
        y_val,
        best_val_prob
    )
)


# Last precision / recall pair has no threshold
precision = precision[:-1]
recall = recall[:-1]


f1_scores = (

    2
    * precision
    * recall

    /

    (
        precision
        +
        recall
        +
        1e-12
    )
)


best_threshold_index = np.argmax(
    f1_scores
)


best_threshold = thresholds[
    best_threshold_index
]


# ============================================================
# 11. FINAL VALIDATION METRICS
# ============================================================

final_val_metrics = evaluate_probabilities(
    y_val,
    best_val_prob,
    best_threshold
)


final_train_metrics = evaluate_probabilities(
    y_train,
    best_train_prob,
    best_threshold
)


print("\n" + "=" * 80)
print("BEST THRESHOLD AFTER RF TUNING")
print("=" * 80)

print(
    f"Threshold:  {best_threshold:.6f}"
)


print("\nTRAIN")

print(
    f"Precision:  {final_train_metrics['precision']:.6f}"
)

print(
    f"Recall:     {final_train_metrics['recall']:.6f}"
)

print(
    f"F1:         {final_train_metrics['f1']:.6f}"
)


print("\nVALIDATION")

print(
    f"Precision:  {final_val_metrics['precision']:.6f}"
)

print(
    f"Recall:     {final_val_metrics['recall']:.6f}"
)

print(
    f"F1:         {final_val_metrics['f1']:.6f}"
)

print(
    f"PR-AUC:     {final_val_metrics['pr_auc']:.6f}"
)

print(
    f"ROC-AUC:    {final_val_metrics['roc_auc']:.6f}"
)


print(
    "\nFinal F1 gap:",
    round(
        final_train_metrics["f1"]
        -
        final_val_metrics["f1"],
        6
    )
)


# ============================================================
# 12. CONFUSION MATRIX
# ============================================================

final_val_pred = (
    best_val_prob >= best_threshold
).astype(int)


tn, fp, fn, tp = confusion_matrix(
    y_val,
    final_val_pred,
    labels=[0, 1]
).ravel()


print("\nConfusion Matrix")

print(f"TP: {tp:,}")
print(f"FP: {fp:,}")
print(f"FN: {fn:,}")
print(f"TN: {tn:,}")


# ============================================================
# 13. FINAL OBJECTS TO KEEP
# ============================================================
#
# best_rf_model
# best_threshold
#
# These become the candidate model + threshold.
#
# DO NOT USE TEST YET.
#
# ============================================================

print("\n" + "=" * 80)
print("TUNING COMPLETE")
print("=" * 80)

print(
    "Keep model as: best_rf_model"
)

print(
    f"Keep threshold as: {best_threshold:.6f}"
)

print(
    "\nTEST SET HAS NOT BEEN USED."
)
```

```python
# ============================================================
# FINAL TEST EVALUATION
# ============================================================
#
# Frozen configuration:
#
# Model:
#     best_rf_model
#
# Features:
#     Random-Forest-selected features
#
# Threshold:
#     best_threshold
#
# IMPORTANT:
#     Nothing is fitted or optimized using y_test.
# ============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    fbeta_score,
    accuracy_score,
    average_precision_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)


# ============================================================
# 1. APPLY THE RF FEATURE SELECTION TO TEST
# ============================================================
#
# rf_mask was learned from TRAIN only.
#
# X_test_processed was transformed using the preprocessor
# already fitted on TRAIN.
# ============================================================

X_test_rf = X_test_processed[:, rf_mask]


print("=" * 70)
print("FINAL TEST SET")
print("=" * 70)

print(
    "Test rows:",
    X_test_rf.shape[0]
)

print(
    "Selected features:",
    X_test_rf.shape[1]
)

print(
    "Frozen threshold:",
    round(float(best_threshold), 6)
)


# ============================================================
# 2. GET TEST PROBABILITIES
# ============================================================

test_probability = best_rf_model.predict_proba(
    X_test_rf
)[:, 1]


# ============================================================
# 3. APPLY FROZEN THRESHOLD
# ============================================================

test_prediction = (
    test_probability >= best_threshold
).astype(int)


# ============================================================
# 4. CALCULATE FINAL METRICS
# ============================================================

test_precision = precision_score(
    y_test,
    test_prediction,
    zero_division=0
)

test_recall = recall_score(
    y_test,
    test_prediction,
    zero_division=0
)

test_f1 = f1_score(
    y_test,
    test_prediction,
    zero_division=0
)

test_f2 = fbeta_score(
    y_test,
    test_prediction,
    beta=2,
    zero_division=0
)

test_accuracy = accuracy_score(
    y_test,
    test_prediction
)

test_pr_auc = average_precision_score(
    y_test,
    test_probability
)

test_roc_auc = roc_auc_score(
    y_test,
    test_probability
)


# ============================================================
# 5. CONFUSION MATRIX
# ============================================================

tn, fp, fn, tp = confusion_matrix(
    y_test,
    test_prediction,
    labels=[0, 1]
).ravel()


# ============================================================
# 6. DISPLAY FINAL RESULTS
# ============================================================

print("\n" + "=" * 70)
print("FINAL TEST PERFORMANCE")
print("=" * 70)

print(
    f"Precision : {test_precision:.6f}"
)

print(
    f"Recall    : {test_recall:.6f}"
)

print(
    f"F1        : {test_f1:.6f}"
)

print(
    f"F2        : {test_f2:.6f}"
)

print(
    f"PR-AUC    : {test_pr_auc:.6f}"
)

print(
    f"ROC-AUC   : {test_roc_auc:.6f}"
)

print(
    f"Accuracy  : {test_accuracy:.6f}"
)


# ============================================================
# 7. CONFUSION MATRIX DETAILS
# ============================================================

print("\n" + "=" * 70)
print("CONFUSION MATRIX")
print("=" * 70)

print(
    f"True Positives  (Fraud detected): {tp:,}"
)

print(
    f"False Positives (False alarms):   {fp:,}"
)

print(
    f"False Negatives (Fraud missed):   {fn:,}"
)

print(
    f"True Negatives  (Correct normal): {tn:,}"
)


# ============================================================
# 8. FRAUD DETECTION RATES
# ============================================================

total_fraud = (
    tp + fn
)

total_non_fraud = (
    tn + fp
)


fraud_detected_percentage = (
    tp / total_fraud * 100
    if total_fraud > 0
    else 0
)

fraud_missed_percentage = (
    fn / total_fraud * 100
    if total_fraud > 0
    else 0
)

false_alarm_percentage = (
    fp / total_non_fraud * 100
    if total_non_fraud > 0
    else 0
)


print("\n" + "=" * 70)
print("BUSINESS INTERPRETATION")
print("=" * 70)

print(
    f"Fraud detected: "
    f"{fraud_detected_percentage:.2f}%"
)

print(
    f"Fraud missed:   "
    f"{fraud_missed_percentage:.2f}%"
)

print(
    f"False alarm rate among legitimate transactions: "
    f"{false_alarm_percentage:.4f}%"
)


# ============================================================
# 9. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        test_prediction,
        target_names=[
            "Non-Fraud",
            "Fraud"
        ],
        digits=4,
        zero_division=0
    )
)


# ============================================================
# 10. COMPARISON WITH VALIDATION
# ============================================================

comparison_df = pd.DataFrame({

    "Metric": [
        "Precision",
        "Recall",
        "F1",
        "PR-AUC",
        "ROC-AUC"
    ],

    "Validation": [
        final_val_metrics["precision"],
        final_val_metrics["recall"],
        final_val_metrics["f1"],
        final_val_metrics["pr_auc"],
        final_val_metrics["roc_auc"]
    ],

    "Test": [
        test_precision,
        test_recall,
        test_f1,
        test_pr_auc,
        test_roc_auc
    ]
})


comparison_df["Difference"] = (
    comparison_df["Test"]
    -
    comparison_df["Validation"]
)


print("\n" + "=" * 70)
print("VALIDATION VS TEST")
print("=" * 70)

display(
    comparison_df
)


# ============================================================
# 11. FINAL MODEL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL MODEL CONFIGURATION")
print("=" * 70)

print("Model:")
print("Random Forest")

print("\nSelected features:")
print(X_test_rf.shape[1])

print("\nThreshold:")
print(round(float(best_threshold), 6))

print("\nParameters:")
print(
    best_rf_model.get_params()
)

print("\nFINAL TEST EVALUATION COMPLETE.")
```

```python
# ============================================================
# LEARNED PREDICTION CONFIDENCE MODEL
# ============================================================
#
# PURPOSE
# -------
# Estimate:
#
#     P(Random Forest prediction is correct | available evidence)
#
# This is DIFFERENT from:
#
#     fraud_probability
#
#
# INPUTS EXPECTED TO ALREADY EXIST
# --------------------------------
#
# best_rf_model
# best_threshold
#
# X_train_rf
# X_val_rf
#
# X_train
# X_val
#
# y_train
# y_val
#
# train_df
#
#
# TEST DATA IS NOT USED.
# ============================================================


import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    brier_score_loss,
    roc_auc_score
)


# ============================================================
# 1. CONFIGURATION
# ============================================================

CONFIDENCE_CONFIG = {

    # Use first 80% of TRAIN time steps to train a temporary RF.
    # Last 20% becomes out-of-sample data for learning confidence.
    "base_train_fraction": 0.80,

    # Historical support reference
    "history_percentile": 95,

    # Confidence labels
    "medium_confidence": 0.90,
    "high_confidence": 0.97,
    "very_high_confidence": 0.995,

    "random_state": 42
}


# ============================================================
# 2. CHECK ALIGNMENT
# ============================================================

assert len(train_df) == X_train_rf.shape[0]
assert len(train_df) == len(X_train)
assert len(train_df) == len(y_train)

assert X_val_rf.shape[0] == len(X_val)
assert X_val_rf.shape[0] == len(y_val)


# ============================================================
# 3. CREATE CHRONOLOGICAL CONFIDENCE-TRAINING SPLIT
# ============================================================
#
# We cannot train the confidence model using predictions that
# the RF made on rows it was trained on.
#
# So inside TRAIN:
#
# Early 80% steps
#       ↓
# temporary RF training
#
# Later 20% steps
#       ↓
# out-of-sample RF predictions
#       ↓
# train confidence model
#
# ============================================================

train_steps = train_df["step"].to_numpy()

unique_steps = np.sort(
    train_df["step"].unique()
)


split_position = int(
    len(unique_steps)
    * CONFIDENCE_CONFIG["base_train_fraction"]
)


confidence_start_step = unique_steps[
    split_position
]


base_mask = (
    train_steps
    < confidence_start_step
)

confidence_train_mask = (
    train_steps
    >= confidence_start_step
)


base_indices = np.where(
    base_mask
)[0]


confidence_indices = np.where(
    confidence_train_mask
)[0]


print("=" * 70)
print("CONFIDENCE MODEL DATA SPLIT")
print("=" * 70)

print(
    "Temporary RF training rows:",
    f"{len(base_indices):,}"
)

print(
    "Confidence training rows:",
    f"{len(confidence_indices):,}"
)

print(
    "Confidence training starts at step:",
    confidence_start_step
)


# ============================================================
# 4. TRAIN TEMPORARY RF
# ============================================================
#
# Same hyperparameters as our selected final RF.
#
# But fit only on EARLY training data.
#
# ============================================================

confidence_base_rf = clone(
    best_rf_model
)


confidence_base_rf.fit(

    X_train_rf[
        base_indices
    ],

    y_train.iloc[
        base_indices
    ]
)


print(
    "\nTemporary Random Forest trained."
)


# ============================================================
# 5. HISTORY FEATURES
# ============================================================

history_features = [

    "customer_previous_transaction_count",

    "merchant_previous_transaction_count",

    "category_previous_transaction_count",

    "customer_merchant_previous_transaction_count",

    "customer_category_previous_transaction_count"
]


# ============================================================
# 6. LEARN HISTORY REFERENCE FROM EARLY TRAIN ONLY
# ============================================================

history_reference = {}


for feature in history_features:

    values = (
        X_train.iloc[
            base_indices
        ][feature]
    )


    reference = np.percentile(

        values,

        CONFIDENCE_CONFIG[
            "history_percentile"
        ]
    )


    history_reference[
        feature
    ] = max(
        float(reference),
        1.0
    )


print("\n" + "=" * 70)
print("HISTORY REFERENCE VALUES")
print("=" * 70)

for feature, reference in history_reference.items():

    print(
        f"{feature:<55}"
        f"{reference:,.2f}"
    )


# ============================================================
# 7. DECISION-MARGIN FUNCTION
# ============================================================

def decision_margin_score(
    probabilities,
    threshold
):

    probabilities = np.asarray(
        probabilities
    )


    score = np.where(

        probabilities >= threshold,

        # Predicted fraud
        (
            probabilities
            - threshold
        )
        /
        (
            1 - threshold
        ),

        # Predicted non-fraud
        (
            threshold
            - probabilities
        )
        /
        threshold
    )


    return np.clip(
        score,
        0,
        1
    )


# ============================================================
# 8. ENTROPY FUNCTION
# ============================================================

def entropy_certainty_score(
    probabilities
):

    p = np.clip(
        np.asarray(probabilities),
        1e-12,
        1 - 1e-12
    )


    entropy = -(

        p * np.log2(p)

        +

        (1 - p)
        * np.log2(1 - p)
    )


    return 1 - entropy


# ============================================================
# 9. BUILD CONFIDENCE FEATURES
# ============================================================

def build_confidence_features(
    model,
    X_rf,
    X_raw,
    threshold,
    history_reference
):

    # --------------------------------------------------------
    # Ensemble fraud probability
    # --------------------------------------------------------

    fraud_probability = (
        model.predict_proba(
            X_rf
        )[:, 1]
    )


    # --------------------------------------------------------
    # Final class decision
    # --------------------------------------------------------

    prediction = (
        fraud_probability
        >= threshold
    ).astype(int)


    # --------------------------------------------------------
    # Decision margin
    # --------------------------------------------------------

    decision_margin = (
        decision_margin_score(
            fraud_probability,
            threshold
        )
    )


    # --------------------------------------------------------
    # Entropy certainty
    # --------------------------------------------------------

    entropy_certainty = (
        entropy_certainty_score(
            fraud_probability
        )
    )


    # --------------------------------------------------------
    # Tree agreement / probability dispersion
    # --------------------------------------------------------

    n_rows = X_rf.shape[0]

    n_trees = len(
        model.estimators_
    )


    agreement_count = np.zeros(
        n_rows,
        dtype=np.int32
    )


    probability_sum = np.zeros(
        n_rows,
        dtype=np.float64
    )


    probability_squared_sum = np.zeros(
        n_rows,
        dtype=np.float64
    )


    for tree in model.estimators_:

        tree_probability = (
            tree.predict_proba(
                X_rf
            )[:, 1]
        )


        # IMPORTANT:
        # Use the SAME operating threshold as the ensemble.
        tree_prediction = (
            tree_probability
            >= threshold
        ).astype(int)


        agreement_count += (
            tree_prediction
            == prediction
        )


        probability_sum += (
            tree_probability
        )


        probability_squared_sum += (
            tree_probability ** 2
        )


    tree_agreement = (
        agreement_count
        /
        n_trees
    )


    tree_mean = (
        probability_sum
        /
        n_trees
    )


    tree_variance = (

        probability_squared_sum
        /
        n_trees

        -

        tree_mean ** 2
    )


    tree_variance = np.maximum(
        tree_variance,
        0
    )


    tree_probability_std = np.sqrt(
        tree_variance
    )


    # --------------------------------------------------------
    # Output feature dataframe
    # --------------------------------------------------------

    confidence_features = pd.DataFrame({

        "fraud_probability":
            fraud_probability,

        "predicted_fraud":
            prediction,

        "decision_margin":
            decision_margin,

        "entropy_certainty":
            entropy_certainty,

        "tree_agreement":
            tree_agreement,

        "tree_probability_std":
            tree_probability_std
    })


    # --------------------------------------------------------
    # Individual historical-support variables
    # --------------------------------------------------------
    #
    # Don't collapse them into one manually weighted number.
    #
    # Let the confidence model learn which history matters.
    #
    # --------------------------------------------------------

    for feature in history_features:

        values = (
            X_raw[feature]
            .to_numpy()
        )


        reference = (
            history_reference[
                feature
            ]
        )


        normalized_support = (

            np.log1p(values)

            /

            np.log1p(reference)
        )


        normalized_support = np.clip(
            normalized_support,
            0,
            1
        )


        confidence_features[
            f"support__{feature}"
        ] = normalized_support


    return (
        confidence_features,
        prediction,
        fraud_probability
    )


# ============================================================
# 10. CREATE OUT-OF-SAMPLE CONFIDENCE TRAINING DATA
# ============================================================

X_confidence_train_raw = (
    X_train.iloc[
        confidence_indices
    ]
    .reset_index(
        drop=True
    )
)


X_confidence_train_rf = (
    X_train_rf[
        confidence_indices
    ]
)


y_confidence_true = (
    y_train.iloc[
        confidence_indices
    ]
    .to_numpy()
)


confidence_train_features, confidence_train_prediction, _ = (
    build_confidence_features(

        confidence_base_rf,

        X_confidence_train_rf,

        X_confidence_train_raw,

        best_threshold,

        history_reference
    )
)


# ============================================================
# 11. CONFIDENCE MODEL TARGET
# ============================================================
#
# 1 = fraud model was correct
# 0 = fraud model was incorrect
#
# ============================================================

confidence_target = (

    confidence_train_prediction
    ==
    y_confidence_true

).astype(int)


print("\n" + "=" * 70)
print("CONFIDENCE TARGET")
print("=" * 70)

print(
    "Correct predictions:",
    f"{confidence_target.sum():,}"
)

print(
    "Incorrect predictions:",
    f"{(confidence_target == 0).sum():,}"
)

print(
    "Base accuracy:",
    f"{confidence_target.mean():.6f}"
)


# ============================================================
# 12. TRAIN CONFIDENCE MODEL
# ============================================================
#
# Logistic Regression gives:
#
# sigmoid(
#       beta0
#     + beta1 * decision_margin
#     + beta2 * tree_agreement
#     + ...
# )
#
# Output:
#
#     P(base prediction is correct)
#
#
# NOTE:
# We intentionally do NOT use class_weight="balanced".
#
# We want the output to behave as a correctness probability,
# not as a class-balanced score.
#
# ============================================================

confidence_model = LogisticRegression(

    max_iter=2000,

    random_state=CONFIDENCE_CONFIG[
        "random_state"
    ]
)


confidence_model.fit(

    confidence_train_features,

    confidence_target
)


print(
    "\nConfidence model trained."
)


# ============================================================
# 13. SHOW LEARNED CONFIDENCE FORMULA
# ============================================================

confidence_coefficients = pd.DataFrame({

    "feature":
        confidence_train_features.columns,

    "coefficient":
        confidence_model.coef_[0]
})


confidence_coefficients[
    "absolute_coefficient"
] = (

    confidence_coefficients[
        "coefficient"
    ].abs()
)


confidence_coefficients = (

    confidence_coefficients
    .sort_values(
        "absolute_coefficient",
        ascending=False
    )
)


print("\n" + "=" * 70)
print("LEARNED CONFIDENCE FACTORS")
print("=" * 70)

display(
    confidence_coefficients
)


print(
    "\nIntercept:",
    confidence_model.intercept_[0]
)


# ============================================================
# 14. BUILD VALIDATION CONFIDENCE FEATURES
# ============================================================
#
# Now use the FINAL frozen Random Forest.
#
# Confidence model itself has NEVER seen validation labels.
#
# ============================================================

validation_confidence_features, val_prediction_conf, val_fraud_probability_conf = (
    build_confidence_features(

        best_rf_model,

        X_val_rf,

        X_val.reset_index(
            drop=True
        ),

        best_threshold,

        history_reference
    )
)


# ============================================================
# 15. ESTIMATE PROBABILITY THAT EACH PREDICTION IS CORRECT
# ============================================================

prediction_confidence = (

    confidence_model
    .predict_proba(
        validation_confidence_features
    )[:, 1]
)


# ============================================================
# 16. ACTUAL VALIDATION CORRECTNESS
# ============================================================

validation_correct = (

    val_prediction_conf
    ==
    y_val.to_numpy()

).astype(int)


# ============================================================
# 17. CREATE FINAL CONFIDENCE OUTPUT
# ============================================================

confidence_results = pd.DataFrame({

    "actual_fraud":
        y_val.to_numpy(),

    "prediction":
        val_prediction_conf,

    "fraud_probability":
        val_fraud_probability_conf,

    # THIS is our new confidence rate:
    "prediction_confidence":
        prediction_confidence,

    "prediction_correct":
        validation_correct
})


# Add diagnostic components

for column in validation_confidence_features.columns:

    confidence_results[
        column
    ] = (
        validation_confidence_features[
            column
        ].to_numpy()
    )


# ============================================================
# 18. CONFIDENCE LABELS
# ============================================================

def get_confidence_label(
    confidence
):

    if (
        confidence
        >= CONFIDENCE_CONFIG[
            "very_high_confidence"
        ]
    ):

        return "Very High"


    if (
        confidence
        >= CONFIDENCE_CONFIG[
            "high_confidence"
        ]
    ):

        return "High"


    if (
        confidence
        >= CONFIDENCE_CONFIG[
            "medium_confidence"
        ]
    ):

        return "Medium"


    return "Low"


confidence_results[
    "confidence_level"
] = [

    get_confidence_label(x)

    for x in prediction_confidence
]


# ============================================================
# 19. OVERALL CONFIDENCE QUALITY
# ============================================================

confidence_brier = brier_score_loss(

    validation_correct,

    prediction_confidence
)


confidence_auc = roc_auc_score(

    validation_correct,

    prediction_confidence
)


print("\n" + "=" * 70)
print("CONFIDENCE MODEL QUALITY")
print("=" * 70)

print(
    f"Brier score: "
    f"{confidence_brier:.6f}"
)

print(
    f"Correctness ROC-AUC: "
    f"{confidence_auc:.6f}"
)


# ============================================================
# 20. VALIDATE CONFIDENCE BANDS
# ============================================================

confidence_summary = (

    confidence_results

    .groupby(
        "confidence_level"
    )

    .agg(

        transactions=(
            "prediction",
            "size"
        ),

        predicted_confidence=(
            "prediction_confidence",
            "mean"
        ),

        actual_accuracy=(
            "prediction_correct",
            "mean"
        ),

        fraud_predictions=(
            "prediction",
            "sum"
        ),

        average_fraud_probability=(
            "fraud_probability",
            "mean"
        )
    )

    .reset_index()
)


confidence_order = [
    "Low",
    "Medium",
    "High",
    "Very High"
]


confidence_summary[
    "confidence_level"
] = pd.Categorical(

    confidence_summary[
        "confidence_level"
    ],

    categories=confidence_order,

    ordered=True
)


confidence_summary = (

    confidence_summary
    .sort_values(
        "confidence_level"
    )
)


print("\n" + "=" * 70)
print("CONFIDENCE VALIDATION")
print("=" * 70)

display(
    confidence_summary
)


# ============================================================
# 21. VALIDATE SEPARATELY FOR FRAUD / NON-FRAUD DECISIONS
# ============================================================

confidence_by_prediction = (

    confidence_results

    .assign(

        prediction_name=lambda df:

        df["prediction"].map({

            0: "Non-Fraud",
            1: "Fraud"
        })
    )

    .groupby(
        [
            "prediction_name",
            "confidence_level"
        ],
        observed=False
    )

    .agg(

        transactions=(
            "prediction_correct",
            "size"
        ),

        predicted_confidence=(
            "prediction_confidence",
            "mean"
        ),

        actual_accuracy=(
            "prediction_correct",
            "mean"
        ),

        average_fraud_probability=(
            "fraud_probability",
            "mean"
        )
    )

    .reset_index()
)


print("\n" + "=" * 70)
print("CONFIDENCE BY PREDICTED CLASS")
print("=" * 70)

display(
    confidence_by_prediction
)


# ============================================================
# 22. SAMPLE FINAL OUTPUT
# ============================================================

print("\n" + "=" * 70)
print("SAMPLE PREDICTIONS WITH CONFIDENCE")
print("=" * 70)


display(

    confidence_results[
        [
            "actual_fraud",
            "prediction",
            "fraud_probability",
            "prediction_confidence",
            "confidence_level",
            "decision_margin",
            "tree_agreement",
            "tree_probability_std",
            "prediction_correct"
        ]
    ]

    .sample(
        20,
        random_state=42
    )

    .sort_values(
        "prediction_confidence",
        ascending=False
    )
)


# ============================================================
# 23. FINAL OBJECTS
# ============================================================
#
# Keep:
#
#     confidence_model
#     history_reference
#     CONFIDENCE_CONFIG
#
#
# For a new transaction:
#
# 1. Base RF gives fraud_probability
# 2. threshold gives fraud / non-fraud decision
# 3. build_confidence_features(...)
# 4. confidence_model.predict_proba(...)[:,1]
#
# gives:
#
#     estimated P(base prediction is correct)
#
# ============================================================

print("\n" + "=" * 70)
print("LEARNED CONFIDENCE LAYER COMPLETE")
print("=" * 70)

print(
    "Confidence now means:"
)

print(
    "Estimated probability that the "
    "Random Forest decision is correct."
)

print(
    "\nTest set was NOT used."
)
```
